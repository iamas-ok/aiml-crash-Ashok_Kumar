```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Function
def f(x):
    return x**2 + 2*x + 1

# Numerical derivative
def derivative(x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)
```

```python
points = [-1, 0, 2]

for p in points:
    print(f"x = {p}")
    print(f"f(x) = {f(p)}")
    print(f"Derivative ≈ {derivative(p):.2f}")
    print()
```

```python
x = np.linspace(-4, 4, 400)
y = f(x)

plt.figure(figsize=(8, 5))
plt.grid(True)
plt.plot(x, y, label='f(x) = x² + 2x + 1')
for p in points:
    slope = derivative(p)
    y_point = f(p)

    tangent = slope * (x - p) + y_point

    plt.plot(x, tangent, '--', label=f'Tangent at x={p}')
    plt.scatter(p, y_point, s=50)
plt.title('Function and Tangent Lines')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```


#### The slope increases as x increases.
#### At x = -1, the slope is nearly zero, indicating the minimum point of the curve.
#### In ML optimization, positive slope means move left, negative slope means move right, and near-zero slope means we are close to an optimum.

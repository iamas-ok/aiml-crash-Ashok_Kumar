```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

uniform_data = np.random.uniform(0, 10, 1000)

normal_data = np.random.normal(5, 2, 1000)
```

```python
uniform_mean = np.mean(uniform_data)
uniform_std = np.std(uniform_data)

normal_mean = np.mean(normal_data)
normal_std = np.std(normal_data)

print("Uniform Distribution")
print("Mean:", round(uniform_mean, 2))
print("Standard Deviation:", round(uniform_std, 2))

print("\nNormal Distribution")
print("Mean:", round(normal_mean, 2))
print("Standard Deviation:", round(normal_std, 2))
```

```python
plt.figure(figsize=(8,5))

plt.hist(uniform_data, bins=20, edgecolor="black")

plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(normal_data, bins=20, edgecolor="black")

plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

#### Observation
##### Uniform distribution has a flat shape because values occur with similar frequency.
##### Normal distribution has a bell-shaped curve with most values near the center.
##### The normal distribution is more concentrated around its mean, while the uniform distribution is spread evenly.

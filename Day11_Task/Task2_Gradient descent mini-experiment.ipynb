```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Loss Function
def loss(theta):
    return (theta - 3.5)**2 + 1

# Gradient
def gradient(theta):
    return 2 * (theta - 3.5)
```

```python
def gradient_descent(start_theta, learning_rate, steps):
    
    results = []

    theta = start_theta

    for i in range(steps):
        grad = gradient(theta)
        current_loss = loss(theta)

        results.append([i, theta, current_loss, grad])

        theta = theta - learning_rate * grad

    return pd.DataFrame(
        results,
        columns=["Iteration", "Theta", "Loss", "Gradient"]
    )
```

```python
start_theta = np.random.uniform(-10, 10)

low_lr = 0.05

low_df = gradient_descent(
    start_theta,
    low_lr,
    15
)

low_df
```

```python
high_lr = 0.5

high_df = gradient_descent(
    start_theta,
    high_lr,
    15
)

high_df
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    low_df["Iteration"],
    low_df["Loss"],
    marker="o",
    label=f"LR = {low_lr}"
)

plt.plot(
    high_df["Iteration"],
    high_df["Loss"],
    marker="s",
    label=f"LR = {high_lr}"
)

plt.title("Loss vs Iteration")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
print("Starting Theta:", start_theta)

print("\nLow Learning Rate Final Loss:")
print(low_df.iloc[-1]["Loss"])

print("\nHigh Learning Rate Final Loss:")
print(high_df.iloc[-1]["Loss"])
```

```python
print("Comparison:")

print("\n1. Low learning rate converges slowly but smoothly.")

print("\n2. High learning rate converges much faster.")

print("\n3. Very high learning rates may overshoot the minimum and become unstable.")
```
